# Violin Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

df=sns.load_dataset("tips")

sns.violinplot(
    data=df,
    x="day",
    y="total_bill"
)
plt.show()
```

# Heatmap

```python
#A heatmap represents values using different intensities
# One of the most important application in Data Science is correlation visualization.
```

```python
# First Calculate Correlation.
correlation = df.corr(numeric_only=True)
print(correlation)
```

```python
sns.heatmap(correlation)
plt.show()
```

```python
sns.heatmap(
    correlation,
    annot=True
)
plt.show()
#annot=True displays the numerical values inside the cells
```

# Pairplot

```python
#One very useful Seaborn Visualization
# It creates multiple plot showing relationship between the numerical variables

sns.pairplot(df)
plt.show()
# diagonal shows the distribution
#off diagonal shows the correlation(relation)
```

```python
sns.pairplot(
    df,
    hue='sex'
)
plt.show()
```

# Distribution Plot

```python
sns.histplot(
    data=df,
    x='total_bill'
)
plt.show()
```

```python
sns.histplot(
    data=df,
    x='total_bill',
    kde=True  #(Kernel Density Estimate) It gives smooth representation of the distribution
)
plt.show()
```

# Customizing Seaborn Appearance

```python
sns.set_theme()
```

```python
sns.set_theme(style="whitegrid") # example of a white grid background
sns.set_theme(style="ticks")
#white
#dark
#Whitegrid
#darkgrid
#ticks
```

```python
sns.histplot(
    data=df,
    x="total_bill"
)
plt.show()
```

# Adding Titles to seaborn Charts

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip"
)
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

# Practice Iris Dataset

```python
import seaborn as sns
df=sns.load_dataset("iris")
print(df)
df["species"].unique()
```

```python
df.info()
```

```python
sns.scatterplot(
    data=df,
    x="sepal_length",
    y="sepal_width"

)
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species"
)
plt.show()
```

```python
sns.lineplot(
    data=df,
    x="species",
    y="sepal_width"
)
plt.show()
```

```python
sns.barplot(
    data=df,
    x="species",
    y="petal_width",
    errorbar=None
)
plt.show()
```

```python
sns.countplot(
    data=df,
    x="species",

)
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="species",
    y="petal_length"
)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="sepal_length"
)

plt.show()
```

```python
sns.violinplot (
    data=df,
    x="species",
    y="sepal_width"
)
plt.show()
```

```python
correlation = df.corr(numeric_only=True)
print(correlation)
```

```python
sns.heatmap(correlation,annot=True)
plt.show()
```

```python
sns.pairplot(df,hue="species")
plt.show()
```

```python
sns.set_theme(style="dark")

fig, ax = plt.subplots()

ax.set_facecolor("black")
fig.patch.set_facecolor("black")

sns.histplot(
    data=df,
    x="sepal_width",
    ax=ax
)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species"
)
plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()
```
