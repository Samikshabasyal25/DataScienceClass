# Dtypes

```python
import numpy as np
a= np.array([10,20,30,40,50])
print(a.dtype)
```

```python
b= np.array([1.5,2.5,3.5])
print(b.dtype)
```

# Itemsize

```python
#Memory size of each element in bytes
a= np.array([10,20,30,40,50])
print(a.itemsize)
```

# nbytes

```python
#total memory used by the array in bytes
a= np.array([10,20,30,40,50])
print(a.nbytes)
```

# Automatic Databytes

```python
import numpy as np
a= np.array([1,2,3.5,6,5.5])
print(a)
print(a.dtype)
```

# Zeros

```python
import numpy as np
a=np.zeros([5])
print(a)
```

# Integer Zeros

```python
import numpy as np
arr=np.zeros(5,dtype=int)
print(arr)
```

```python
#2d zeros
import numpy as np
arr=np.zeros((3,4))
print(arr)
```
