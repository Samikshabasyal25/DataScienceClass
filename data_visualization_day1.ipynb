# Adding Grid

```python
import matplotlib.pyplot as plt
months=['January','February','March','April','May','June','July','August','September','October','November','December']
sales =[1000,1200,1500,1300,1700,1600,1800,2000,2200,2100,2300,2500]

plt.figure(figsize=(12,8))
plt.plot(months,sales,marker='o',linestyle='-',color='g')
plt.title('Monthly Sales Data')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid()
plt.show()
```

# Multiple Lines

```python
months=['January','February','March','April','May','June','July','August','September','October','November','December']
product_a =[1000,1200,1500,1300,1700,1600,1800,2000,2200,2100,2300,2500]
product_b=[2400,2300,2200,2000,2100,1500,1400,1200,800,700,900,1000]

plt.figure(figsize=(12,8))
plt.plot(months,product_a,marker='o',linestyle='-',color='g',label='Product A')
plt.plot(months,product_b,marker='s',linestyle='--',color='r',label='Product B')
plt.title('Monthly Sales Data for Product A and B')

plt.xlabel('Month')
plt.ylabel('Sales')

plt.legend()
plt.show()
```

# Bar Graph

```python
students = ['Alice', 'Bob', 'Charlie', 'David', "Eva"]
marks=[85,92,78,90,88]
plt.bar(students,marks,color='lightblue',edgecolor='black')

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
students = ['Alice', 'Bob', 'Charlie', 'David', "Eva"]
marks=[85,92,78,90,88]
plt.barh(students,marks,color='lightblue',edgecolor='black')

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

# Histogram

```python
marks=[85,92,78,90,88,95,80,88,85,90,87,90,94,84,92,90]
plt.hist(marks,bins=range(70,105,5),color='lightgreen',edgecolor='black')

plt.title("Distribution of Student Marks")
plt.xlabel('Marks')
plt.ylabel('Frequency')

plt.show()
```
