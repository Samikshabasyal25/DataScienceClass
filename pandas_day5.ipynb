# Clean the given data

```python
import pandas as pd
import numpy as np
```

```python
data ={
    'Customer':["Alice", "Bob","Charlie", "Alice", "Eve","Frank",None],
    'Age':[25,None,35,40,'Thirty',29,33],
    'City':['NYC','new york','New York','LA','Los Angeles','la','NYC'],
    'Purchase':[250,300,None,400,500,600,250],
    'Gender':['F','M','Male','M','F','Female',None], 
}
```

```python
df= pd.DataFrame(data)
print("Uncleaned Data")
print(df)
```

```python
print(df.isnull())
```

```python
df['Purchase']=df['Purchase'].fillna(df['Purchase'].mean())
print(df)
```

```python
df['Age']=df['Age'].replace({'Thirty':30})
df['Age']=pd.to_numeric(df['Age'])
df['Age']=df['Age'].fillna(df['Age'].mean())
print(df)
```

```python
df=df.dropna()
```

```python
df['City']=df['City'].str.lower().replace({
    'nyc': 'New York',
    'LA' : 'Los Angeles',
    'la' : 'Los Angeles',
    'los angeles' : 'Los Angeles',
    'new york' :'New York'
})
print(df)
```

```python
df['Gender']= df['Gender'].replace({
    'Male':'M',
    'Female':'F'
})
```

```python
df=df.drop_duplicates()
```

```python
print('Cleaned Data')
print(df)
```

```python
df.to_csv('Cleaned_data.csv',index=True)
```

```python
import pandas as pd
df=pd.read_csv('students.csv')
print("Uncleaned Data")
print(df)
```

```python
print(df.isnull())
```

```python
df.loc[df['Age']<0,'Age']=None
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Marks']=df['Age'].fillna(df['Marks'].mean())
print(df)
```

```python
df['Name']=df['Name'].str.strip().str.title()
df['City']=df['City'].str.strip().str.title()
print(df)
```

```python
df['Gender']=df['Gender'].replace({
    'Male':'M',
    'Female':'F',
    'male':'M',
    'female':'F',
    'Male':'M',
    'Unknown':'M'
})
print(df)
```

```python
df['Email']=df['Email'].fillna(df['Name'].str.lower()+'@gmail.com')
print(df)
```

```python
df['Join_Date']= pd.to_datetime(df['Join_Date'],errors='coerce')
df= df.dropna(subset=['Join_Date']).reset_index(drop=True)
print(df)
```

```python
df_clean = df.copy()
duplicate_count = df_clean.drop(columns=['ID']).duplicated().sum()
print('Duplicate rows:', duplicate_count)
```

```python
cols_to_check = df.columns.drop('ID')

df_no_dup = (
    df.drop_duplicates(subset=cols_to_check, keep='first')
      .reset_index(drop=True)
)

print('Final rows:', len(df_no_dup))
```

```python
df_no_dup.to_csv('Cleaned_studentsdata.csv', index=False)
```
