# Data Transformation

```python
data={
    "Name":['Alice','Bob','Charlie','David','Eve'],
    "Marks":[80, 90, 75, 85, 95]
}
df=pd.DataFrame(data)
print(df)
```

```python
#add percentage column
df['Percentage']= df['Marks']/100 * 100
print(df)
```

```python
#increment marks by 5%
df['Marks']= df['Marks']*1.05
print(df)
```

# GroupBy

```python
data={
    'dept':['HR','IT','Finance','Marketing','Sales'],
    'salary':[5000,7500,6000,7800,4500]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.describe())
```

```python
print(df.groupby('dept')['salary'].mean())
```

```python
df.groupby('Gender')['Age'].sum()
```

# Sorting

```python
data={
   "Name":['Alice','Bob','Charlie','David','Eve'],
   'Age':[25,30,35,40,45],
   'City':["Butwal","Bhairahawa","kathmandu","Newyork","Chicago"],
    "Marks":[80, 90, 75, 85, 95] 
}
df=pd.DataFrame(data)
print(df)
```

```python
#sort by ascending order of marks
print(df.sort_values("Marks",ascending=True))
```

```python
print(df.sort_values("Age",ascending=False))
```

```python
print(df.sort_values("City",ascending=True))
```
