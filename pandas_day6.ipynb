```python
import pandas as pd
```

```python
df=pd.read_csv('employee_data.csv')
print("Uncleaned Data")
print(df)
```

```python
print(df.isnull())
```

```python
df.info()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
df.loc[df['Salary']<0,'Salary']=None
df['Salary']=df['Salary'].fillna(df['Salary'].mean())
print(df)
```

```python
df['Gender']=df['Gender'].str.title()
df['City']=df['City'].str.strip().str.title()
df['Name']=df['Name'].str.strip().str.title()
print(df)
```

```python
df.info()
```

```python
df = df[df['Email'].str.contains('@', na=False)]
df=df.dropna()
print(df)
```

```python
df['Join_Date'] = pd.to_datetime(
    df['Join_Date'],
    errors='coerce',
    format='mixed'
)
df['Join_Date']=df['Join_Date'].dt.strftime('%Y-%m-%d')
print(df)
```

```python
df.loc[df['Age']>100,'Age']=None
df['Age']=df['Age'].fillna(df['Age'].mean())

df['Age']=df['Age'].round().astype(int)
pd.set_option('display.float_format','{:.5f}'.format)
print(df)
```

```python
df_clean = df.copy()
duplicate_count = df_clean.drop(columns=['ID']).duplicated().sum()
print('Duplicate rows:', duplicate_count)
```

```python
cols_to_check = df.columns.drop('ID')

df_no_dup = (
    df.drop_duplicates(subset=cols_to_check, keep='first')
      .reset_index(drop=True)
)

print('Final rows:', len(df_no_dup))
```

```python
df_no_dup.to_csv('Cleaned_employee_data.csv', index=False)
```
