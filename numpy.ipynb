# Numpy array

```python
import numpy as np
numbers=np.array([1,2,3,4,5])
print(numbers)
print(type(numbers))
```

# ndarray

```python
# 0 dimensional array
zero_dimensional_array = np.array(42)
print(zero_dimensional_array)
```

```python
#1 dimensional array
one_dimensional_array = np.array([1,2,3,4,5])
print(one_dimensional_array)
```

```python
# 2 dimensional array
two_dimensional_array = np.array([[1,2,3],[4,5,6]])
print(two_dimensional_array)
```

```python
# 3 dimensional array
#collection of matrices
three_dimensional_array = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(three_dimensional_array)
```

# Integer Zeros

```python
import numpy as np
arr=np.zeros(5,dtype=int)
print(arr)
```

```python
#2d zeros
import numpy as np
arr=np.zeros((3,4))
print(arr)
```

# np.ones

```python
#create an array filled with ones
import numpy as np
arr= np.ones(5,dtype=int)
print(arr)
```

```python
import numpy as np
arr= np.ones((2,2))
print(arr)
```

# np.empty

```python
#creates an array without initializing its element
import numpy as np
arr= np.empty(5)
print(arr)
```

```python
import numpy as np
arr= np.empty(5)
arr[:]=[1,2,3,4,5]
print(arr)
```

# np.full()

```python
#creates an array where every element has the same value,provided by the user
import numpy as np
arr= np.full((3,3),7)
print(arr)
```

# np.eye()

```python
#creates a identical like matrix where the main diagonal is 1 and all other element are 0
import numpy as np
arr= np.eye(5)
print(arr)
```

```python
import numpy as np
arr= np.eye(3,5)
print(arr)
```
